# Uber Data Analysis — Summary of Work Completed

## 1. Dataset Overview

**Analysis Performed:**

- Examined the dataset to understand its structure, including the total number of records and columns.
- Explored ride IDs, customer IDs, driver IDs, dates, fares, ratings, vehicle types, and cancellation details.
- Performed basic statistical analysis on numerical and categorical columns.

**Key Findings:**

- The dataset contains 80,000 ride records and 20 columns.
- The ride dates cover the period from January 1 to January 9, 2023.
- The dataset contains several operational fields that can be used to analyze ride trends and service performance.

**Identified Issues and Recommendations:**

- Some numerical values, particularly distance, ride duration, and fare, appear unusually large.
- The measurement units and dataset generation method should be verified before using these values for financial or operational decisions.

## 2. Data Quality and Cleaning

**Analysis Performed:**

- Checked for missing values.
- Identified duplicate rows and duplicate ride IDs.
- Validated distance, duration, fare, ratings, and driver experience values.
- Checked the formats and validity of the date and time columns.

**Key Findings:**

- Duplicate rows: 0
- Duplicate ride IDs: 0
- Invalid distance, duration, fare, and rating values: 0
- Missing values in `cancellation_reason`: 71,956
- Missing values in `promo_code`: 64,012

**Identified Issues and Recommendations:**

- Missing cancellation reasons for completed rides were replaced with `Not Applicable`, as the cross-tabulation showed that all missing cancellation reasons belonged to completed rides.
- Missing promo codes were not filled because these values could indicate either that no promotion was used or that the information was not recorded.
- The meaning of missing promo codes should be confirmed using the dataset's source documentation before further cleaning.

## 3. Ride Status Analysis

**Analysis Performed:**

- Calculated the number and percentage of completed and cancelled rides.
- Compared ride statuses with their recorded cancellation reasons.

| Ride Status | Number of Rides | Percentage |
|---|---:|---:|
| Completed | 71,956 | 89.95% |
| Cancelled | 8,044 | 10.06% |
| **Total** | **80,000** | **100%** |

**Key Findings:**

- The dataset contains significantly more completed rides than cancelled rides.
- Approximately 10.06% of all recorded rides were cancelled.

**Identified Issues and Recommendations:**

- To understand and potentially reduce cancellations, factors such as cancellation reasons, waiting times, pickup locations, and the booking experience should be investigated.
- Comparing cancellation rates across dates, locations, and vehicle types could help identify areas with higher cancellation rates.

## 4. Cancellation Reason Analysis

**Analysis Performed:**

- Analyzed cancelled rides based on their recorded cancellation reasons.

| Cancellation Reason | Number of Rides | Percentage of Cancellations |
|---|---:|---:|
| Rider Cancelled | 4,005 | 49.79% |
| Driver Cancelled | 2,451 | 30.47% |
| No Show | 1,588 | 19.74% |
| **Total** | **8,044** | **100%** |

**Key Findings:**

- Rider cancellations represent the largest cancellation category.
- Driver cancellations are the second-largest category.
- No-show cases also account for a meaningful proportion of total cancellations.

**Identified Issues and Recommendations:**

- Possible factors behind rider cancellations include long waiting times, fare expectations, or changes in travel plans.
- Possible factors behind driver cancellations include pickup distance, driver availability, or traffic conditions.
- The dataset does not clearly identify whether the rider or driver was absent in no-show cases.
- The company could collect more detailed cancellation feedback, improve arrival notifications, and record no-show reasons separately.

*Note: These are possible explanations based on the observed patterns. The dataset does not confirm the exact causes of cancellations.*

## 5. Date and Time Analysis

**Analysis Performed:**

- Examined the data types of the `ride_date` and `ride_time` columns.
- Checked for invalid dates and times.
- Identified the date range and the number of unique calendar days.
- Compared the time components of both columns.

**Key Findings:**

- The dataset contains 9 unique calendar dates.
- Invalid dates: 0
- Invalid times: 0
- A total of 79,997 records have mismatches between the time components of `ride_date` and `ride_time`.
- All records in `ride_date` have zero minutes and seconds, whereas `ride_time` contains more precise time values.

**Identified Issues and Recommendations:**

- The intended meaning of both columns needs to be confirmed.
- The original columns should be preserved while the source documentation is reviewed.
- Once the column definitions are verified, reliable daily, hourly, and time-of-day analyses can be performed.

## 6. Daily Ride Trend Analysis

**Analysis Performed:**

- Calculated the total number of rides for each date.
- Created a line chart to visualize daily ride trends.
- Compared completed and cancelled rides across dates.
- Examined ride times and ride status distributions for January 9.

**Key Findings:**

- January 3 recorded the highest number of rides, with 9,825 rides.
- January 5 recorded the lowest number of rides among the first eight days, with 9,475 rides.
- Daily ride volumes remained relatively stable from January 1 to January 8.
- The total number of recorded rides dropped sharply to 3,137 on January 9.
- On January 9, 2,823 rides were completed and 314 were cancelled.
- Records were distributed across the 24-hour period on January 9; however, this alone does not confirm that the day's data is complete.

**Identified Issues and Recommendations:**

- The sharp decline on January 9 could indicate an actual reduction in demand or incomplete data.
- Date-wise data completeness and source records should be verified before interpreting this decline as a genuine business trend.
- Once the data has been validated, the results may support demand forecasting and daily operational planning.

## 7. Driver ID Analysis

**Analysis Performed:**

- Calculated the total number of unique driver IDs and ride IDs.
- Examined sample driver IDs and their frequency in the dataset.

**Key Findings:**

- Unique driver IDs: 80,000
- Unique ride IDs: 80,000
- Sample driver IDs follow a sequential pattern, such as `D00001`, `D00002`, and so on.

**Identified Issues and Recommendations:**

- Having a unique driver ID for every ride is unusual in a typical ride-hailing operation, where drivers generally complete multiple rides.
- The method used to generate driver IDs should be verified.
- Active driver counts and individual driver performance should not be interpreted until the meaning of these IDs is confirmed.

## 8. Overall Business Recommendations

Based on the analysis completed so far, the following actions are recommended:

1. Investigate cancellation reasons in greater detail and monitor cancellation rates regularly.
2. Verify the meaning of missing values and promo code records.
3. Confirm the definitions and intended use of the date and time columns.
4. Validate the unusually low ride count recorded on January 9 against the source data.
5. Verify the driver ID generation pattern and the measurement units used for numerical values.
6. In the next phase, investigate relationships involving vehicle types, pickup and drop-off locations, fare patterns, customer and driver ratings, traffic conditions, and weather conditions.

## 9. Conclusion

The initial analysis identified important patterns in ride status distribution, cancellation reasons, daily ride trends, and data quality. Most rides were recorded as completed, while rider cancellations represented the largest category of cancellations.

At the same time, several observations require further verification, including inconsistencies between date and time fields, unusually large numerical values, and the presence of a unique driver ID for every ride.

Resolving these data-related concerns will improve the reliability of the analysis and provide a stronger foundation for further investigation into vehicle performance, operational efficiency, and potential business improvement opportunities.
